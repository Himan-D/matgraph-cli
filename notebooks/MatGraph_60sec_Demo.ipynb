# MatGraph in 60 seconds — no API key needed

This notebook runs entirely on **bundled demo data**: real crystal structures and X-ray diffraction patterns, plus illustrative ML samples. Nothing to configure — just run the cells.

````python
# In Colab, install first (uncomment the next line):
# %pip install -q matgraph-cli

from matgraph import MatGraphSDK

sdk = MatGraphSDK(demo=True)  # zero-config: no API key, no PyTorch needed
print("MatGraph demo SDK ready. Formulas:", ["LiFePO4", "Si", "NaCl"])
````

## 1. Predict material properties (sample)

````python
rows = sdk.predict("LiFePO4", model="m3gnet")
r = rows[0]
print("Formula:      ", r["formula"], "(", r["crystal_system"], ")")
print("Pred E_form:  ", r["predicted_form_energy"], "eV/atom  [SAMPLE]")
print("Pred energy:  ", round(r["m3gnet_energy"], 3), "eV  [SAMPLE]")
print("Density:      ", r["features"]["density"], "g/cm^3  [real crystallography]")
print("Provenance:   ", r["provenance"]["mode"])
````

## 2. Simulate an X-ray diffraction pattern (real crystallography)

````python
xrd = sdk.xrd("Si")
peaks = sorted(zip(xrd["two_theta"], xrd["intensity"], xrd["hkls"]), key=lambda p: -p[1])
print('2-theta   I(%)  hkl')
for t, i, h in peaks:
    print(f'{t:7.2f}  {i:6.1f}  {h[0]}')
# Si 111 reflection at 28.44 deg Cu-Ka — check a crystallography textbook.
````

## 3. Triage a hypothetical compound (sample)

````python
res = sdk.substitute("LiFePO4", "Li", "Na")
print(res["original"]["formula"], round(res["original"]["energy"], 3), "eV")
print(res["hypothetical"]["formula"], round(res["hypothetical"]["energy"], 3), "eV")
print("More stable?", res["is_more_stable"], "[SAMPLE — validate candidates with DFT]")
````
